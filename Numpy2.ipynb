## An array of random numbers 
-  random array is an array containing values generated randomly using NumPy’s random module.
- * np.random.rand() → random decimal numbers
- * np.random.randint() → random integer numbers

```python
# random
# randn
# randint
```

```python
import numpy as np
```

```python
np.random.random(10) # Always Positive between 0 to 1
```

```python
np.random.randint(1,100,5) # Need to give range and how many elements you want
```

```python
np.random.randn(10) # Random values between -3 to +3
```

```python
# 2)Substraction
l1 = np.array([1,2,3,4])
l2 = np.array([11,12,13,14])
l1-l2
```

## Operations on a 2D Array

- 1) Addition
- 2) Multiplication
- 3) Comparision
- 4) matrix multiplication
- 5) Transpose

```python
l1 = np.arange(1,5,1).reshape(2,2)
l2 = np.arange(5,9,1).reshape(2,2)
print(l1)
print(l2)
```

```python
l1+l2
```

```python
l1-l2
```

```python
l1/l2
```

```python
l1*l2
```

```python
l1@l2 # Matrix Multiplication
```

```python
np.matmul(l1,l2) # Matrix Multiplication
```
