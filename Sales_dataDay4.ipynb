```python
import pandas as pd
sales=pd.read_csv('https://raw.githubusercontent.com/tkseneee/Dataset/master/k_circle_sales.csv')
```

```python
cat_data = sales.select_dtypes(include='str')
num_data = sales.select_dtypes(exclude='str')
```

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
scaled_weight = sc.fit_transform(num_data)
num_scaled_data = pd.DataFrame(scaled_weight, columns=num_data.columns)
num_scaled_data
```

```python
num_scaled_data.plot(kind='kde')
```

```python
import seaborn as sns

sns.scatterplot(x=num_scaled_data['Item_Weight'],y=num_scaled_data['Item_Visibility'])
```

```python
import matplotlib.pyplot as plt
for i in num_scaled_data.columns:
               sns.kdeplot(data=num_scaled_data,x=i)
               plt.show()
```

```python
# normalization
# value ranges from 0 to 1 
# min max scaling

# xi-xmin/xmax-xmin
from sklearn.preprocessing import MinMaxScaler

sc=MinMaxScaler()
scaled_weight = sc.fit_transform(num_data)
num_scaled_data = pd.DataFrame(scaled_weight,columns=num_data.columns)
num_scaled_data
```

```python
num_scaled_data.describe().round()
```

```python
for i in num_scaled_data.columns:
               sns.kdeplot(data=num_scaled_data,x=i)
               plt.show()
```

```python
# robust scalar
# xi-median/iqr

from sklearn.preprocessing import RobustScaler

sc = RobustScaler()
scaled_weight= sc.fit_transform(num_data)
num_scaled_data = pd.DataFrame(scaled_weight,columns=num_data.columns)
num_scaled_data
```

```python
for i in num_scaled_data.columns:
               sns.kdeplot(data=num_scaled_data,x=i)
               plt.show()
```

```python
# NaN dtype
# 1) dropping approach
#         dropping the column/rows
#         if the missing the data is more or above 40% of the data
# 2) imputing approach

#         numeric 
#         mean :- If the data is missing around 3%
#         median :- If the data is missing around 3%
#         group by mean and median if the missing data around 5%
#         ML based imputations


#         Category
#         Mode
#         rule based imputation
```
