```python
import pandas as pd
sales=pd.read_csv('https://raw.githubusercontent.com/tkseneee/Dataset/master/k_circle_sales.csv')
```

```python
sales
```

```python
sales["Profit_Category"] = pd.cut(
    sales["Profit"],
    bins=[0, 10, 12, 14, 16],
    labels=["Low", "Medium", "High", "Very High"]
)
```

```python
sales
```

```python
sales["MRP_Visibility"] = (sales["Item_MRP"] * sales["Item_Visibility"]).round()
```

```python
sales
```

```python
sales["Sales_Performance"] = pd.qcut(
    sales["Item_Outlet_Sales"],
    q=4,
    labels=["Poor", "Average", "Good", "Excellent"]
)
```

```python
sales
```

```python
import numpy as np
age = np.array([6,5,4,3,2,6,5,100])
age.mean()
```

```python
from scipy import stats

age.sort()
stats.trim_mean(age,proportiontocut=0.2)
```

```python
cat_data = sales.select_dtypes(include='object')
num_data = sales.select_dtypes(exclude='object')
```

```python
propcut = stats.trim_mean(num_data,proportiontocut=.10)
result=pd.DataFrame()

result['mean']=num_data.mean()
result['median']=num_data.median()
result['trim_mean']=propcut
result
```
