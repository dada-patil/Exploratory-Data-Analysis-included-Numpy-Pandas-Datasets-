## Pivot Table

```python
import pandas as pd

d = {
    'Name': ['Jony', 'Anil', 'Sunil', 'Jony', 'Anil', 'Rocky',
             'Jony', 'Anil', 'Sunil', 'Jony', 'Anil', 'Rocky'],

    'Exam_Months': ['3_Months', '3_Months', '3_Months', '3_Months', '3_Months', '3_Months',
                    '6_Months', '6_Months', '6_Months', '6_Months', '6_Months', '6_Months'],

    'Subject': ['Python', 'Python', 'SQL', 'SQL', 'ML', 'ML',
                'Python', 'Python', 'SQL', 'SQL', 'ML', 'ML'],

    'Theory': [62, 47, 55, 74, 31, 77, 85, 63, 42, 67, 89, 81],

    'Lab': [70, 60, 63, 80, 40, 77, 67, 70, 80, 90, 92, 95]
}

df = pd.DataFrame(
    d,
    columns=['Name', 'Exam_Months', 'Subject', 'Theory', 'Lab']
)

df

```

```python
# Syntax :-
# pd.pivot_table(dataframe,index=columns,aggfunc=np.mean)

# recommandation

# index = non-numerical column
```

```python
# in agg_func - np.mean , np.median , np.std , np.var ,np.max , np.min
```

```python
import numpy as np
pd.pivot_table(df,index="Subject",aggfunc = np.min)
```

```python
pd.pivot_table(df,index="Subject",aggfunc = np.max)
```

```python
pd.pivot_table(df,index="Subject",values ="Lab",aggfunc = np.mean)
```

```python
df[df["Subject"]=="Python"]["Lab"].mean()
```

```python
df[df["Subject"]=="ML"]["Lab"].mean()
```

```python
df[df["Subject"]=="SQL"]["Lab"].mean()
```

```python
pd.pivot_table(df,index=["Subject"],values =["Lab","Theory"],aggfunc = np.mean)
```

```python
df.groupby(["Exam_Months", "Subject"])[["Lab", "Theory"]].mean()
```

```python

pd.pivot_table(df,index=["Exam_Months","Subject"],values =["Lab","Theory"],aggfunc = np.mean)
```

```python
pd.pivot_table(df,index=["Exam_Months","Subject"],values =["Lab","Theory"],aggfunc = np.sum)
```

```python
df.groupby(["Exam_Months","Subject"]).agg({"Lab":np.mean,"Theory":np.sum})
```

```python
df.groupby(["Exam_Months","Subject"]).agg(np.sum)
```

```python
lab = df.groupby(["Exam_Months"]).agg({"Lab":np.mean})
lab
Th = df.groupby(["Exam_Months"]).agg({"Theory":np.mean})
Th
avg = lab*Th
print(avg)
```

```python
df.groupby(["Exam_Months","Subject"]).agg({"Lab":np.mean,"Theory":np.std})
```
