## Univariate analysis

```python
# Condition for normal distribution
# mean = median = mode
# 1.68% data lies between 1 sigma (std) 
# 2.95.5% data lies between 2 sigma (std)
# 3.99.7% of the data
# skewness = 0
# if skewness normal then the data will be looking bell shape curved
# skewness formula E(xi-Xbar)**3/n/std**3
# Kurtosis measures the shape/tailedness and peakedness of a distribution, particularly how heavy or light its tails are relative to a normal distribution.
```

```python
num_data['Item_Outlet_Sales'].plot(kind='kde',color="purple")
```

```python
num_data['Profit'].plot(kind='kde',color='orange')
```

```python
num_data['Item_MRP'].plot(kind='kde')
```

```python
num_data.skew()
# if the values is near to zero the its normal distribution
# if the value is in negative side then left skewed data
# if the values is in positive side then right skewed data
```

```python
num_data.skew().plot(kind='kde',color='red')
```

```python
# kurstosis
# * Mesokurtic: Similar to normal distribution. near to zero
# * Leptokurtic: Heavier tails and more extreme values. / have heavy outliers greater than zero
# * Platykurtic: Lighter tails and fewer extreme values. / some outliers less than zero
num_data.kurt()
```

```python
num_data.kurt().plot(kind='kde',color='red')
```
