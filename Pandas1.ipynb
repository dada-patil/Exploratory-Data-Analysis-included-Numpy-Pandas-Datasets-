### Creating Pandas Series

```python
ser = pd.Series([10,20,40,77,99,80])
print(ser)
```

```python
# Customized index

ser = pd.Series([10,20,40,77,99,80], index=['a', 'b', 'c', 'd', 'e', 'f'])
print(ser)
```

```python
# recomended to use the default index but if need create a index using the string 
```

```python
print(ser.size)  # Total no of elements
print(ser.shape) # size of the series
print(ser.dtype) # Data type of the series
```

### Adding and Replacing elements

```python
ser = pd.Series(["one","two","three","four"])
ser
```

```python
ser[4]="five" # Add
ser
```

```python
ser[3]="FOUR" # Update
ser
```

```python
## add elements
## variable[new_index]=value

## Modify the values
## variable[existing_index]=new value
```

### Deleting elements
- Syntax:
- ser_name.drop(index=[index of values need to be dropped])

```python
ser.drop(index=[4])
```

```python
ser[2]="hundred"
ser[5]="hundred"
ser
```

```python
ser.drop(index=[2,5],inplace=True) # inplace = True is permanataly deleted and inplace=False is temorary deleted
ser
```

```python
# Change is happening but the changes are not saved in the series

# Syntax :- series.reset_index(drop=True)
```

```python
ser.reset_index(drop=True)
```

```python
ser
```

```python
ser.reset_index(drop=True,inplace =True)
ser
```

```python
ser.reset_index(drop=False)
```
