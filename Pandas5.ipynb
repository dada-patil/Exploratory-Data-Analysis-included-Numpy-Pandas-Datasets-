## Conditional formatting or Filtering

```python
# Syntax :-
# df[col] - condition - boolean def
# df[df[col] - condition] - filter the df
```

```python

name = ["Ram", "john", "Abdul", "Siva", "Adam", "Rahman"]
maths = [45, 40, 60, 80, 90, 75]
english = [60, 75, 90, 65, 80, 60]

di = {
    'names': name,
    'Maths': maths,
    'English': english
}

scores_df = pd.DataFrame(di)
print(scores_df)
```

```python
scores_df['Maths']>60
```

```python
scores_df[scores_df['Maths']>60]
```

```python
scores_df['English']>=60
```

```python
scores_df[scores_df['English']>60]
```

```python
# Syntax :-

# df[(condition1)logicalsymbol(cond2)]
# only logical symbol needs to be used 
# each cond needs to be separated by brackets
```

```python
scores_df[scores_df['English']>60][["names","Maths"]]
```

```python
scores_df[(scores_df['Maths']>=60) & (scores_df['English']>=70)]
```

```python
scores_df[(scores_df['Maths']>=60) & (scores_df['English']>=70)][["names","Maths"]]
```

```python
scores_df[(scores_df['Maths']>=60) & (scores_df['English']>=70)][["names","English"]]
```

```python
scores_df[(scores_df["Maths"]>=60)|(scores_df["English"]>=70)]
```

```python
scores_df[scores_df["names"]=="john"]
```

```python
scores_df[scores_df["names"]=="john"][["names","English"]]
```

```python
scores_df[scores_df["names"]!="john"]
```

```python
import pandas as pd


name = ["Ram", "john", "Abdul", "Siva", "Adam", "Rahman"]
maths = [45, 40, 60, 80, 90, 75]
english = [60, 75, 90, 65, 80, 60]

di = {
    'names': name,
    'Maths': maths,
    'English': english
}

scores_df = pd.DataFrame(di)
print(scores_df)
```

```python
# Syntax :-

# df['colname'] = df['colname'].replace({oldvalue:newvalue}) # To replace in the
# df['rowname'] = df['rowname'].replace({oldvalue:newvalue})
```

```python
scores_df['Maths'] = scores_df['Maths'].replace({40:42}) # Replace maths column 40 to 42
scores_df
```

```python
scores_df['English']= scores_df['English'].replace({65:67}) # replace English column 65 to 67
scores_df
```

```python
scores_df
```

```python
scores_df.replace({75:77},inplace= True)
scores_df
```

```python
scores_df.loc[3,"English"]= [70]
```

```python
scores_df
```
