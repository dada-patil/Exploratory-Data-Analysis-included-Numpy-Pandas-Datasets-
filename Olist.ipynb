```python
# Delivery time
items['delivery_days'] = (
    items['order_delivered_customer_date']
    - items['order_purchase_timestamp']
).dt.total_seconds() / (24 * 60 * 60)
items['delivery_days'].describe()
```

```python
# Estimated Delivery Time
items['estimated_delivery_days'] = (
    items['order_estimated_delivery_date']
    - items['order_purchase_timestamp']
).dt.total_seconds() / (24 * 60 * 60)
items['estimated_delivery_days']
```

```python
# Delivery Delay
items['delivery_delay_days'] = (
    items['order_delivered_customer_date']
    - items['order_estimated_delivery_date']
).dt.total_seconds() / (24 * 60 * 60)
items['delivery_delay_days']
```

```python
# Late Delivery flag
items['late_delivery'] = np.where(
    items['delivery_delay_days'] > 0,
    'Late',
    'On Time'
)
items['late_delivery']
```

```python
items['late_delivery'].value_counts()
```

```python
# Late delivery Percentage
late_delivery_rate = (
    items['late_delivery']
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

late_delivery_rate
```

```python
delivered_orders = items[
    items['order_delivered_customer_date'].notna()
].copy()

late_rate = (
    delivered_orders['delivery_delay_days'].gt(0).mean() * 100
)

round(late_rate, 2)
```

```python
# Merge orders+items
order_value = (
    orders.groupby('order_id')
    .agg(
        product_value=('price', 'sum'),
        freight_value=('freight_value', 'sum'),
        number_of_items=('order_item_id', 'count')
    )
    .reset_index()
)
```

```python
# Create Total order value
order_value['total_order_value'] = (
    order_value['product_value']
    + order_value['freight_value']
)
order_value['total_order_value'] 
```

```python
order_value.head()
```

```python
# Merge with orders
order_analysis = items.merge(
    order_value,
    on='order_id',
    how='left'
)
```

```python
# freight burden
#order_analysis['freight_value'] = order_analysis['freight_value_x']
order_analysis['freight_percentage'] = (
    (order_analysis['freight_value']
    / order_analysis['product_value']
) * 100).round(2)
```

## Review Analysis

```python
reviews.head()
```

```python
reviews['review_score'].value_counts().sort_index()
```

```python
# Review score distribution
sns.countplot(
    data=reviews,
    x='review_score',
    order=sorted(reviews['review_score'].dropna().unique())
)

plt.title("Customer Review Score Distribution")
plt.xlabel("Review Score")
plt.ylabel("Number of Reviews")
plt.show()
```

```python
# Review score percentage
review_pct = (
    reviews['review_score']
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
)

review_pct
```

```python
# Review score average
reviews['review_score'].mean()
```

```python
reviews['review_score'].median()
```

```python
# Create review category
reviews['review_category'] = np.where(
    reviews['review_score'] >= 4,
    'Positive',
    np.where(
        reviews['review_score'] == 3,
        'Neutral',
        'Negative'
    )
)
```

```python
reviews['review_category']
```

```python
reviews['review_category'].value_counts()
```

```python
# Review category percentage
reviews['review_category'].value_counts(normalize=True).mul(100).round(2)
```

```python
# merge reviews with orders
order_reviews = items.merge(
    reviews[['order_id', 'review_score', 'review_category']],
    on='order_id',
    how='left'
)
```

```python
order_reviews
```

```python
# Does late delivery relate to customer satisfaction
late_review = (
    order_reviews
    .groupby('late_delivery')['review_score']
    .agg(['count', 'mean', 'median'])
    .round(2)
)

late_review
```

```python
# Review distribution :late vs on time
review_late = pd.crosstab(
    order_reviews['late_delivery'],
    order_reviews['review_score'],
    normalize='index'
) * 100

review_late.round(2)
```

```python
review_late.plot(
    kind='bar',
    figsize=(10,5)
)

plt.title("Review Score Distribution by Delivery Performance")
plt.xlabel("Delivery Performance")
plt.ylabel("Percentage")
plt.legend(title="Review Score")
plt.show()
```
