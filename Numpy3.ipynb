## Concatenation of an Array

### Vertical Concatenaton
- Syntax :- np.concat(list/tuple of array,axis=0)
- np.vstack(list of array)

- condition:
- 1)column count needs to be equal

```python
import numpy as np
arr1 =np.arange(1,10,1).reshape(3,3)
arr2 = np.arange(11,20,1).reshape(3,3)
print(arr1)
print(arr2)
```

```python
np.concatenate([arr1,arr2],axis=0)
```

```python
np.vstack([arr1,arr2])
```

### Horizontal Concatenation

- Syntax :- np.concat(list/tuple of array,axis=1)
- np.hstack(list of array)

- condition:
- 1)rows count needs to be equal


```python
np.concatenate([arr1,arr2],axis=1)
```

```python
np.hstack([arr1,arr2])
```

```python
# practice :
# 1) Conatenate 2*3 , 4*3 array  using axis = 0
arr1 = np.arange(1,7,1).reshape(2,3)
arr2 = np.arange(11,23,1).reshape(4,3)
np.concatenate([arr1,arr2],axis=0)
```

```python
#2 Conatenate 2*3 , 2*3 array  using axis = 1
arr1 = np.arange(1,7,1).reshape(2,3)
arr2 = np.arange(11,17,1).reshape(2,3)
np.concatenate([arr1,arr2],axis=1)
```

## Split

#### hsplit - draw a vertical line to split them equally
#### vsplit - draw a horizontal line to split them equally

- Syntax : np.vsplit(array,no of equal split)
- Syntax : np.hsplit(array,no of equal split)

```python
arr1 = np.arange(1,10,1).reshape(3,3)
np.vsplit(arr1,3)
```

```python
np.hsplit(arr1,3)
```

```python
arr2 = np.arange(1,17,1).reshape(4,4)
arr2
```

```python
np.vsplit(arr2,2)
```

```python
np.vsplit(arr2,4)
```

```python
np.hsplit(arr2,2)
```

```python
np.hsplit(arr2,4)
```

### Iterating Array

- Syntax : np.nditer(array_name)

```python
arr1 = np.arange(1,10)
print(arr1)
for i in np.nditer(arr1):
    print(i)
```

```python
a1 = np.arange(1,10).reshape(3,3)

for i in np.nditer(a1):
               print(i)
```

```python
a1 = np.arange(1,10).reshape(3,3)

for i in a1:
               print(i)
```

```python
a1 = np.arange(1,10).reshape(3,3)

for i in np.nditer(a1):
               if i%2 == 0:
                              print("Even :",i)
               else:
                              print("Odd :",i)
```

```python
# Create 2D array of random numbers and print even and odd numbers from it.
a1 = np.random.randint(1,10,9).reshape(3,3)
print(a1)
for i in np.nditer(a1):
               if i%2 == 0:
                              print("Even :",i)
               else:
                              print("Odd :",i)
```

```python
# Create 2D array of random numbers and print even and odd numbers from it.
a1 = np.random.randint(1,10,size=(3,3,3))

for i in np.nditer(a1):
               if i%2 == 0:
                              print("Even :",i)
               else:
                              print("Odd :",i)
```
