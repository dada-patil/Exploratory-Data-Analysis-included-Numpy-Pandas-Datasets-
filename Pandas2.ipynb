## Data Frame

- creating Dataframe

```python
# key = col_names
# list of values = elements of each column
# Syntax :
# data_frame_name = pd.DataFrame(dictionary/list of list,index(opt),columns(opt))
```

```python
import pandas as pd
```

```python
emp_data = pd.DataFrame({"emp_id":[101,102,103,104,105],
              "name":["suriya","john","doe","paul","karthick"],
              "dept":["AIML","ML","DA","DS","PD"],
              "salary":[20,30,24,45,50]})
emp_data
```

```python
emp_id = pd.Series([101,102,103,104,105])
name = pd.Series(["suriya","john","doe","paul","karthick"])
dept = pd.Series(["AIML","ML","DA","DS","PD"])
salary = pd.Series([20,30,24,45,50])

pd.DataFrame([emp_id,name,dept,salary],index=['emp_id','name','dept','salary']).T
```

- iloc =  index location

```python
# Syntax:-
# df_name.iloc[row_index,col_index]

scores_df.iloc[0,0]
```

```python
scores_df.iloc[1,0]
```

```python
scores_df.iloc[2,0]
```

```python
scores_df.iloc[2,2]
```

```python
scores_df.iloc[[0,3,5],[0,1,2]] # .iloc[[row],[col]]
```

```python
# slicing

scores_df.iloc[2:5:2,0:3:1] # .iloc[row:row:step,col:col:step]
```

```python

scores_df.iloc[0:,0:3:1]  # Even rows
```

```python
scores_df.iloc[::2,::2] 
```

```python
scores_df.iloc[::2,:] 
```

```python
scores_df.iloc[:,::2] 
```

```python
# list of list Row wise
scores_df = pd.DataFrame([["Ram",45,60],
                          ["john",40,75],
                          ["Abdul",60,90],
                          ["Siva",80,65],
                          ["Adam",90,80],
                          ["Rahman",75,60]],
                         columns = ["Student_name","Maths_marks","English_marks"])
scores_df
```

```python
# practice


name = ["Ram", "john", "Abdul", "Siva", "Adam", "Rahman"]
maths = [45, 40, 60, 80, 90, 75]
english = [60, 75, 90, 65, 80, 60]

di = {
    'names': name,
    'Maths': maths,
    'English': english
}

scores_df = pd.DataFrame(di)
print(scores_df)
```

```python
print(scores_df.iloc[1,1])
```

```python
print(scores_df.iloc[[1,2,3],[1]])
```

```python
print(scores_df.iloc[[1,3,4],[1,2]])
```

- loc = name based indexing

```python
# Syntax:-
# df_name.loc[indexname,columnname]

scores_df.loc[0,"names"]
```

```python
print(scores_df.loc[0,"Maths"])
```

```python
print(scores_df.loc[0,"English"])
```

```python
print(scores_df.loc[[0,1,4],["names","Maths","English"]])
```

```python
# slicing. here the (name type) value in included
print(scores_df.loc[0:4:1,"names":"Maths"])
```

```python
# Mixed Indexing and slicing
print(scores_df.loc[0:4:1,["names","English"]])
```
